```python
# Node class with constructor:
class Node:
    def __init__(self, data):
        self.data=data
        self.next=None
```

```python
# Add few elements
node1 = Node("Extract Data")
node2 = Node("Transform Data")
node3 = Node("Load Data")


# Link the above objects to each other
head = node1
current=head
node1.next = node2
node2.next = node3
node3.next = None
```

```python
# print the order

print("#### Start ####")
while current:
    print(current.data, end=" -> ")
    current = current.next
print("\n#### End ####")
```

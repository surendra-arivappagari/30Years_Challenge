# Datetime Package:

#### In Python real-world projects, the `datetime` package is very useful for all date, time, timezone specific value preparations. Mainly we do have 5 different flavours of datetime been used. 
- 1.`date`:
    - Year, Month, Day only.	Birthdays, Daily report dates.
- 2.`time`:
    - Hour, Minute, Second, Microsecond.	Event triggers within a single day.
- 3.`datetime`:
    - Combines date + time.	The Gold Standard. Logs, DB timestamps.
- 4.`timedelta`:
    - Represents a Duration (Difference).	Finding "Records older than 30 days."
- 5.`timezone`:
    - Handles UTC offsets.	Global data pipelines (in all organizations, this process been followed rigorously due to multi-regional workspaces.

### 1.Date: `datetime.date`

```python
#Prepare today, find min, max values from the module. 

from datetime import date
today=date.today()

print("Today date is: ", today)

print("Minimum date=", date.min)

print("Maximum date=", date.max)
```

```python
#create date object using values:
today=date(2025, 12, 31)
print("Given date= ", today)

print("\nGiven year= ",today.year)
print("Given month= ",today.month)
print("Given day= ",today.day)
```

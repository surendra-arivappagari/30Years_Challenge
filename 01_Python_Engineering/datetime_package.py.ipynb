# Datetime Package:

#### In Python real-world projects, the `datetime` package is very useful for all date, time, timezone specific value preparations. Mainly we do have 5 different flavours of datetime been used. 
- 1.`date`:
    - Year, Month, Day only.	Birthdays, Daily report dates.
- 2.`time`:
    - Hour, Minute, Second, Microsecond.	Event triggers within a single day.
- 3.`datetime`:
    - Combines date + time.	The Gold Standard. Logs, DB timestamps.
- 4.`timedelta`:
    - Represents a Duration (Difference).	Finding "Records older than 30 days."
- 5.`timezone`:
    - Handles UTC offsets.	Global data pipelines (in all organizations, this process been followed rigorously due to multi-regional workspaces.

<hr>

### 1.Date: `datetime.date`

```python
#Prepare today, find min, max values from the module. 

from datetime import date
today=date.today()

print("Today date is: ", today)

print("Minimum date=", date.min)

print("Maximum date=", date.max)
```

```python
#create date object using values:
today=date(2025, 12, 31)
print("Given date= ", today)

print("\nGiven year= ",today.year)
print("Given month= ",today.month)
print("Given day= ",today.day)
```

<hr>

### 2.Time: `datetime.time`

```python
from datetime import time
current_time=time(12,56,59)
print("Current time=", current_time)

print("\nCurrent time hours=", current_time.hour)
print("Current time minute=", current_time.minute)
print("Current time seconds=", current_time.second)
```

### 3.DateTime: `datetime.datetime`

```python
#Get the timestamp

from datetime import datetime
timestamp = datetime.now()
print("current timestamp=", timestamp)
```

### 4.TimeDelta: datetime.timedelta¶

Timedelta is widely used for adding subtracting given intervals for any timestamp or time or date fields. 
- days, seconds, minutes, hours, weeks:

```python
from datetime import datetime, timedelta
current_time = datetime.now()

print("1). Before Addd current_time=", current_time)

add_2_hours=current_time+timedelta(hours=2)
print("2). Aftet Addd 2 hours to current_time=", add_2_hours)

add_5_days=add_2_hours+timedelta(days=5)
print("2). Aftet Addd 2 hours to current_time=", add_5_days)
```

```python
from datetime import datetime, timezone
import pytz

Current_timestamp=datetime.now()
print("Current_timestamp=", Current_timestamp)

UTC_Current_timestamp=datetime.now(pytz.utc)
print("\nUTC_Current_timestamp=", UTC_Current_timestamp)

IST_Current_timestamp=datetime.now(pytz.timezone('Asia/Kolkata'))
print("\nIST_Current_timestamp=", IST_Current_timestamp)


US_Current_timestamp=datetime.now(pytz.timezone('US/Central'))
print("\nUS_Current_timestamp=", US_Current_timestamp)
```
